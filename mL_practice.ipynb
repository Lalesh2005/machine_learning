```python
def remove_sentence(text, target):
    if target in text:
        modified = text.replace(target, "")
        modified = " ".join(modified.split())
        return modified
    return text

if __name__ == "__main__":
    input_file = "input.txt"
    target_file = "target.txt"
    output_file = "output.txt"

    with open(input_file, "r", encoding="utf-8") as f:
        content = f.read()

    with open(target_file, "r", encoding="utf-8") as f:
        target = f.read().strip()

    modified_content = remove_sentence(content, target)

    with open(output_file, "w", encoding="utf-8") as f:
        f.write(modified_content)

    print(f"Done. Modified content written to {output_file}")
```

```python
def extended_gcd(p,q):
    if p == 0:
        return (q, 0, 1)
    else:
        (gcd, u, v) = extended_gcd(q % p, p)
        return (gcd, v - (q // p) * u, u)
p = 26513
q = 32321
gcd, u, v = extended_gcd(p, q)

print(u,v)
```

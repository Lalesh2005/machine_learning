## Central Tendency -- Mean,Median,Mode

```python
## Mean
import numpy as np
list = [1,2,3,4,5,5]
np.mean(list)
```

```python
## Median
np.median(list)
```

```python
## Mode
values, counts = np.unique(list, return_counts=True)
mode = values[np.argmax(counts)]
print(mode)
```

## Dispersion

```python
# Range
range_val = np.max(list) - np.min(list)

# Variance
variance = np.var(list)

# Standard Deviation
std_dev = np.std(list)

print(range_val)
print(variance)
print(std_dev)
```

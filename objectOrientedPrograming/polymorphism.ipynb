### Polymorphism


```python
## Methods overriding in Python
## Base class

class Animal:
    def speak(self):
        return "sound of the animal"

## Derived class 1
class Dog(Animal):
    def speak(self):
        return "Woof!"

## Derived class 
class Cat(Animal):
    def speak(self):
        return "Meow!"
    
## function that demonstrate plymorphism

def animal_speak(animal):
    print(animal.speak())

dog=Dog()
cat=Cat()
print(dog.speak())
print(cat.speak())
animal_speak(dog)
```

```python
### Ploymorphism with function and methods
## base class
class shape:
    def area(self):
        return "The area of the figure"

## Derived class1  
class Rectangle(shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height
    
    def area(self):
        return self.width*self.height
    
## Derived class2

class Circle(shape):
    def __init__(self,radius):
        self.radius=radius
    
    def area(self):
        return 3.14*self.radius*self.radius

## Function that demonstrate polymorphism

def print_area(shape):
    print(f"the area is {shape.area()}")
    
rectangle=Rectangle(4,5)
circle=Circle(3)

print_area(rectangle)
print_area(circle)


```

```python
## Polymorphism with abstract base class
from abc import ABC,abstractmethod
## Define abstract class 
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
    
## derived class 1 
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"

##derived class 2

class Motorcyle(Vehicle):
    def start_engine(self):
        return "motorcycle engine started"
    
def start_vehicle(Vehcile):
    print(Vehcile.start_engine)
## Create objects of the car and motorcycele 
car=Car()
motorcycle=Motorcyle()
start_vehicle(car)
```

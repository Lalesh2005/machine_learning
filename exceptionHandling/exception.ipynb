```python
## try except else 
try:
    num=int(input("Enter a number: "))
    result=10/num
except ZeroDivisionError:
    print("Cannot divide by zero.")
except ValueError:  
    print("Invalid input. Please enter a valid integer.")
else:  
    print(f"Result is {result}")
```

```python
## try except else finally
try:
    num=int(input("Enter a number: "))
    result=10/num
except ZeroDivisionError:
    print("Cannot divide by zero.")
except ValueError:  
    print("Invalid input. Please enter a valid integer.")
else:  ## if error has not occured it has to executed
    print(f"Result is {result}")
finally: ## weather the error is coming or not it has to be excuted
    print("Execution completed.")
```

```python
import requests
import pandas as pd


# Apna URL yahan daalo
URL = "https://samagama.in/api/matrix-mystics/leaderboard?limit=5000"

# Request bhejo
response = requests.get(URL)
#response.raise_for_status()

# JSON data lo
data = response.json()

print(type(data))
print(data)

# Leaderboard list nikalo
leaderboard = data.get("leaderboard", [])

# Required fields extract karo
rows = []

for person in leaderboard:
    rows.append({
        "Rank": person.get("rank"),
        "Name": person.get("name"),
        "LinkedIn": person.get("linkedin")
    })

# DataFrame banao
df = pd.DataFrame(rows)

# Excel file me save karo
output_file = "leaderboard.xlsx"
df.to_excel(output_file, index=False)

print(f"Saved {len(df)} records to {output_file}")
```

```python
import browser_cookie3
import requests
import pandas as pd

# Chrome se cookies load karo
cj = browser_cookie3.chrome(domain_name="samagama.in")

# API call
response = requests.get(
    "https://samagama.in/api/matrix-mystics/leaderboard?limit=5000",
    cookies=cj,
)

print("Status:", response.status_code)

response.raise_for_status()

data = response.json()

# Structure verify karne ke liye
print("Keys:", data.keys())

leaderboard = data.get("leaderboard", [])

rows = []

for person in leaderboard:
    rows.append({
        "Rank": person.get("rank"),
        "Name": person.get("name"),
        "LinkedIn": person.get("linkedin"),
    })

df = pd.DataFrame(rows)

# Rank ke hisab se sort
df = df.sort_values("Rank")

# Excel save
output_file = "leaderboard.xlsx"
df.to_excel(output_file, index=False)

print(f"Saved {len(df)} records to {output_file}")
print(df.head())
```

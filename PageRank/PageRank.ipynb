```python
df = pd.read_csv("endorsement_network.csv")
G = nx.from_pandas_edgelist(
    df,
    source="endorser",
    target="endorsed",
    edge_attr="weight",
    create_using=nx.DiGraph()
)
```

```python
print(G)
```

```python
pagerank_scores = nx.pagerank(
    G,
    alpha=0.85,
    weight="weight"
)
```

```python
top_pr = (
    pd.DataFrame(
        pagerank_scores.items(),
        columns=["Person","PageRank"]
    )
    .sort_values(
        "PageRank",
        ascending=False
    )
)

print(top_pr.head())
```

```python
in_degree = dict(G.in_degree())
top10_indegree = (
    pd.DataFrame(
        in_degree.items(),
        columns=["Person","InDegree"]
    )
    .sort_values(
        "InDegree",
        ascending=False
    )
    .head(10)
)

print(top10_indegree)
```

## 7. Damping Factor

```python
for alpha in [0.5,0.85,0.95]:

    pr = nx.pagerank(
        G,
        alpha=alpha,
        weight="weight"
    )

    top10 = (
        pd.DataFrame(
            pr.items(),
            columns=["Person","PR"]
        )
        .sort_values(
            "PR",
            ascending=False
        )
        .head(10)
    )

    print(f"\nAlpha={alpha}")
    print(top10)
```

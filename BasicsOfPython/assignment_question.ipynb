**Question 18:** Write a Python program to reverse a string.

```python
# Your code here
a = "Lalesh"
b = len(a)
c = ""   # Initialize c before using it

for i in range(1, b + 1):
    c = c + a[b - i]   # concatenate characters from end
print(c)
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
# Your code here
a="abba"
b = len(a)
c = ""   # Initialize c before using it

for i in range(1, b + 1):
    c = c + a[b - i]   # concatenate characters from end
if(c==a):
    print("yes")
```

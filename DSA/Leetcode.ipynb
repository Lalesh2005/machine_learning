```python
# Largest in Array
def largest(nums):
        """
        :type nums: List[int]
        :rtype: str
        """
        max=nums[0]
        for num in nums:
            if (max<=num):
                max=num
        return max
nums = [10,2]
print(largest(nums))
```

```python
#1752. Check if Array Is Sorted and Rotated
def check(nums):
        """
        :type nums: List[int]
        :rtype: bool
        """
        count =0
        n=len(nums)
        for i in range(0,n-1):
            if(nums[i]>(nums[i+1]%n)):
                count=count+1
            
        if count<=1:
            return True
        return False
            
nums = [3,4,5,1,2]
print(check(nums))
```

```python
#389. Find the Difference
def findTheDifference(s, t):
        """
        :type s: str
        :type t: str
        :rtype: str
        """
        freq = {}
        for ch in s:
            if ch in freq:
                freq[ch]+=1
            else:
                freq[ch]=1
        for ch in t:
            if ch in freq and freq[ch]!=0:
                freq[ch]-=1
            else:
                return ch
        
s = "a"
t = "aa"
findTheDifference(s, t)
```

```python
#451. Sort Characters By Frequency
def frequencySort(s):
        """
        :type s: str
        :rtype: str
        """
        freq = {}
        for ch in s:
            if ch in freq:
                freq[ch] += 1
            else:
                freq[ch] = 1
        word = ""
        freq = sorted(freq.items(), key=lambda x: x[1], reverse=True)
        for key, count in freq:   # <-- tuple unpacking
                word += key*count
        return word
s = "tree"
frequencySort(s)
```

```python
#344. Reverse String
def reverseString(s):
        """
        :type s: List[str]
        :rtype: None Do not return anything, modify s in-place instead.
        """
        t=[]
        for i in range(len(s)-1,-1,-1):
            t.append(s[i])
        for i in range(0,len(s)):
            s[i]=t[i]
        print(s)
s = ["h","e","l","l","o"]
reverseString(s)
```

# decorators-- ye na kisi function ko extend karne ke kaam aaata hai 
Decorator ek function hai jo kisi dusre function ke around wrap hokar usme extra functionality add karta hai bina original function ko modify kiye.
-- without decorator code:

* def my_decorator(func):
*    def wrapper():
*        print("Before function")
*        func()
*        print("After function")
*    return wrapper
* def greet():
*    print("Hello")
* greet = my_decorator(greet)
* greet()

-- using decorator code

* @my_decorator
* def greet():
*    print("Hello")
* greet()

```python
def deco(func):

    def wrapper():
        print("A")
        func()
        print("B")

    return wrapper


@deco
def hello():
    print("C")

hello()
```

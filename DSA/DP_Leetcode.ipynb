```python
## Recurcive solution of fibonacci
## 0 1 2 3 5 8 13 
def fibo(n):
    if(n<=1):
        return n
    return fibo(n-1)+fibo(n-2)
fibo(5)
```

```python
## Top-Down approach 
def fibo(n,dp):
    if(n<=1):
        return n
    if(dp[n]!=-1):
        return dp[n]
    
    dp[n]=fibo(n-1,dp)+fibo(n-2,dp)
    return dp[n]
n=100
dp = [-1]*(n+1)
fibo(n,dp)
```

```python
# Bottom Up 
def fibo(n,dp):
    if n<=1:
        return n
    dp[0]=0
    dp[1]=1
    for i in range(2,n+1):
        dp[i]=dp[i-1]+dp[i-2]
    return dp[n]


n=100
dp = [-1]*(n+1)
fibo(n,dp)
```

```python
## space optimization 
def fibo(n,dp):
    if n<=1:
        return n
    dp[0]=0
    dp[1]=1
    for i in range(2,n+1):
        dp[2]=dp[0]+dp[1]
        dp[0]=dp[1]
        dp[1]=dp[2]
    return dp[2]


n=5
dp = [-1]*(3)
fibo(n,dp)
```

```python
#509. Fibonacci Number
def fib(n):
        """
        :type n: int
        :rtype: int
        """
        if n<=1:
            return n
        dp = [-1]*3
        dp[0]=0
        dp[1]=1
        for i in range(2,n+1):
            dp[2]=dp[0]+dp[1]
            dp[0]=dp[1]
            dp[1]=dp[2]
        
        return dp[2]
fib(4)
```

```python
#1137. N-th Tribonacci Number
def tribonacci(n):
        """
        :type n: int
        :rtype: int
        """
        if n<=1:
            return n
        if n==2:
            return 1

        dp = [-1]*4
        dp[0]=0
        dp[1]=1
        dp[2]=1
        for i in range(3,n+1):
            dp[3]=dp[0]+dp[1]+dp[2]
            dp[0]=dp[1]
            dp[1]=dp[2]
            dp[2]=dp[3]

        return dp[3]

tribonacci(25)

        
```

```python
def minEnergy(index,heights,k):
    if index == 0:
        return 0
    result = float('inf')
    for i in range(1, min(k, index) + 1):
    # code
        result = min(result,abs(heights[index]-heights[index-i])+minEnergy(index-i,heights,k))

    return result 


def minimizeEnergy(heights,k):
    n=len(heights)
    return minEnergy(n-1,heights,k)

heights = [10,20,30,10]
k=2
minimizeEnergy(heights,k)
```

```python
#78. Subsets
def count(nums,index,result,curr):
        if index == len(nums):
            result.append(curr[:])
            return
        # include
        curr.append(nums[index])
        count(nums,index+1,result,curr)
        # backtrack
        curr.pop()
        # element ko exclude karo
        count(nums, index + 1,result,curr)
        
        
def subsets(nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        result = []
        count(nums,0,result,[])
        return result 

nums = [1,2,3]
subsets(nums)
```

```python
with open('example.txt','r') as file:
    content=file.read()
print(content)
```

```python
## read file line by line
with open('example.txt','r') as file:
    for line in  file:
        print(line.strip()) ## strip() removes the newlinw charcter
```

```python
## writing of file(overwriting)
with open('example.txt','w') as file:
    file.write('Hello world\n')
    file.write('this is new line')
```

```python
### write a file(without overwriting)
with open('example.txt','a') as file:
    file.write("append operation takig place!")
```

```python
## writing a list of lines to a file
lines=['first line \n','second line \n','third line\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
## writing a list of line to a files
data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
## Reading a binary file 
with open('example.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
## read the content from the source text file and write it 
## to a destination text file 
## copying  a text file

with open('example.txt','r') as source_file:
    content=source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
```

```python
## writing and then reading a file
with open('example.txt','w+') as file:
    file.write("Hello world!")
    file.write("this is a new line\n")
    
    ## the file cursor to the begning 
    file.seek(0)
    
    ## read the content of the file 
    content=file.read()
    print(content)
```

```python
## checking if a path is a file or a directory 
import os 
path ='example.txt'
if os.path.isfile(path):
    print(f"The path '{path}' is a file")
elif os.path.isdir(path):
    print(f"The path '{path}' is dir")
else:
    print("the path is neither dir neither file")
```

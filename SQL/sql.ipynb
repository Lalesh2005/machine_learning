```python
import sqlite3
```

```python
## connect to an sqlite database 
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
## create a table 
cursor.execute('''
              create Table if Not Exists employees(
                  id Integer Primary Key,
                  name Text Not null,
                  Age Integer,
                  Department text
              )
               ''')

## comit the changes
connection.commit()
```

```python
cursor.execute('''
               select * from employees
               ''')
```

```python
## insert the data in sqlite table
cursor.execute('''
                Insert Into employees(name,Age,Department)
                values('LALESH',23,'Btech')
               ''')

cursor.execute('''
                Insert Into employees(name,Age,Department)
                values('SUMAN',23,'Biology')
               ''')
cursor.execute('''
                Insert Into employees(name,Age,Department)
                values('RAMAN',23,'MATHEMATICS')
               ''')

connection.commit()
```

```python
## Query the data from the table 
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## Print the quired data
for row in rows:
    print(row)
```

```python
## Upadte the data int the table 
cursor.execute('''
               Update employees
               Set Age=24
               Where name="LALESH"
               
               ''')
connection.commit()
```

```python
## DElete the data from the table 
cursor.execute('''
               Delete from employees
               where name="LALESH"
               ''')
connection.commit()
```

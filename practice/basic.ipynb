## 1. FizzBuzz ⭐⭐⭐⭐⭐

Problem

1 se N tak print karo:

* Multiple of 3 → Fizz
* Multiple of 5 → Buzz
* Multiple of both → FizzBuzz

```python
n = int(input("Enter the number"))
for i in range(1,n+1):
    if i%3==0 and i%5==0:
        print("FizzBuzz")
    elif i%5==0:
        print("Buzz")
    elif i%3==0:
        print("Fizz")
    else:
        print(i)
    
```

## 3. Armstrong Number ⭐⭐⭐⭐

Example 153 
output  true
Because 1³ + 5³ + 3³ = 153

```python
n = int(input("Enter the number"))
temp = n
sum=0
while(temp>0):
    c =temp%10
    sum=(c**3)+sum
    temp=int(temp/10)
if(sum==n):
    print("true")
else:
    print("false")
```

## 5. Factorial Without Recursion ⭐⭐⭐⭐⭐

Example 5
output 120

```python
def fact(n):
    if(n<=0):
        return -1
    if(n==1):
        return 1
    return fact(n-1)*n

fact(20)
```

6. Pattern Printing (Star Pyramid) ⭐⭐⭐⭐⭐

    *
   ***
  *****
 *******

```python
n = 4

for i in range(1, n + 1):

    # spaces
    for j in range(n - i):
        print(" ", end="")

    # stars
    for k in range(2 * i - 1):
        print("*", end="")

    print()
```

## 7. Guess the Number Game ⭐⭐⭐⭐⭐

Problem

Computer generates a number.

User keeps guessing until correct.

Output:Too High
Too Low
Correct

```python
import numpy as np

n = np.random.randint(1,100)

low = 1
high = 100

while True:

    guess = np.random.randint(low, high)

    print(f"Guess = {guess}")

    if guess == n:
        print("Got it!")
        break

    elif guess < n:
        print("Too Low")
        low = guess + 1

    else:
        print("Too High")
        high = guess
```

## 4. Character Frequency Count

```python
s ="akbbaaccde"

freq={}
for ch in s:
    if ch in freq:
        freq[ch]+=1
    else:
        freq[ch]=1
print(freq)
```

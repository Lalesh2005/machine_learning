```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Bias values (probability of heads)
p = np.linspace(0, 1, 100)
```

```python
# Avoid log(0)
epsilon = 1e-10
entropy = - (p + epsilon)*np.log2(p + epsilon) - (1 - p + epsilon)*np.log2(1 - p + epsilon)
```

```python
gini = 2 * p * (1 - p)
```

```python
# Plot
plt.figure()
plt.plot(p, entropy, label="Entropy")
plt.plot(p, gini, label="Gini Impurity")

plt.xlabel("Bias (Probability of Heads)")
plt.ylabel("Impurity / Uncertainty")
plt.title("Entropy vs Gini Impurity (Coin Toss)")
plt.legend()
plt.grid()

plt.show()
```

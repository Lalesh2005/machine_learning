```python
## applying label encoding on the sex and income
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df_copy['sex']=le.fit_transform(df_copy['sex'])
df_copy['income']=le.fit_transform(df_copy['income'])
```

```python
df_copy.head()
```

```python
## applying One-Hot encoding for the 'workclass', 'occupation', 'relationship', 'race', 'native-country'
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
```

```python
encoder.fit_transform(df_copy[['workclass']]).toarray()
```

```python
new_workclass=pd.DataFrame(encoder.fit_transform(df_copy[['workclass']]).toarray(),columns=encoder.get_feature_names_out())
```

```python
df_copy = pd.concat([df_copy, new_workclass], axis=1)
df_copy.head()
```

```python
df_copy.drop(axis=1,columns='workclass',inplace=True)
```

```python
df_copy.head()
```

```python
## now apply OneHotEncoder for the rest 
cols = ['occupation', 'relationship', 'race', 'native-country','marital-status']

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

# Fit + Transform all columns together
encoded = encoder.fit_transform(df_copy[cols])

# Create DataFrame with proper column names
encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(cols)
)

# Concatenate with original dataframe
df_copy = pd.concat([df_copy, encoded_df], axis=1)

# Drop original columns
df_copy.drop(columns=cols, inplace=True)

df_copy.head()
```

```python
df_copy.shape
```

```python
df_copy.drop('education',axis=1,inplace=True)
```

```python
df_copy.to_csv('Salary_pred_cleaned_dataset.csv')
```

## Applying Decision Treee

```python
new_df=pd.read_csv('Salary_pred_cleaned_dataset.csv')
new_df.head()
```

```python
new_df.columns
```

```python
new_df.shape
```

```python
X = new_df.drop(columns='income')
```

```python
X.head()
```

```python
X.shape
```

```python
Y=new_df['income']
```

```python
Y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

## DECISION TREE

```python
from sklearn.tree import DecisionTreeClassifier
treeClassifier = DecisionTreeClassifier(max_depth=2)
```

```python
treeClassifier.fit(X_train,Y_train)
```

```python
from sklearn import tree 
plt.figure(figsize=(20, 10))
tree.plot_tree(treeClassifier,filled=True)
```

```python
Y_pred=treeClassifier.predict(X_test)
from sklearn.metrics import confusion_matrix,classification_report
```

```python
cm=confusion_matrix(Y_test,Y_pred)
print(cm)
print(classification_report(Y_test,Y_pred))
```
